```python
import torch

last_token_logits = torch.tensor([-1, 2, -3, 3, -3, 0, 2, 1])
temperature = 0.7
p = 0.95

dist = torch.nn.functional.softmax((last_token_logits / temperature))
probs, tokens = torch.sort(dist, descending=True)
c_probs = torch.cumsum(probs, dim=0)

keep = (c_probs - probs) < p 
tokens = tokens[keep]
print(tokens)
```
